# Expressões Regulares (RegEx) em Python

---

# O que é RegEx?

**RegEx (Regular Expression)** é uma sequência de caracteres que define um padrão de busca.

Ela é utilizada para:

- Encontrar textos
- Validar entradas
- Extrair informações
- Substituir partes de uma string
- Dividir textos

Exemplos:

- Encontrar todos os e-mails
- Validar CPF
- Encontrar telefones
- Encontrar datas
- Renomear arquivos

---

# Importando o módulo

```python
import re
```

Todo o trabalho com expressões regulares acontece através do módulo **re**.

---

# Principais funções

## re.search()

Procura o padrão em qualquer lugar da string.

```python
import re

texto = "The rain in Spain"

resultado = re.search("Spain", texto)

print(resultado)
```

Saída

```
<re.Match object>
```

Se não encontrar:

```
None
```

---

## re.findall()

Retorna **todas** as ocorrências.

```python
import re

texto = "banana"

print(re.findall("a", texto))
```

Resultado

```
['a', 'a', 'a']
```

---

## re.split()

Divide uma string usando um padrão.

```python
import re

texto = "maça,banana,uva"

print(re.split(",", texto))
```

Resultado

```
['maça', 'banana', 'uva']
```

---

## re.sub()

Substitui um padrão.

```python
import re

texto = "Python é legal"

novo = re.sub("legal", "incrível", texto)

print(novo)
```

Resultado

```
Python é incrível
```

---

```python
resultado = re.search("Spain", texto)
```

Podemos acessar:

```python
resultado.start()
```

Posição inicial.

```python
resultado.end()
```

Posição final.

```python
resultado.span()
```

Tupla com início e fim.

```python
resultado.group()
```

# Strings Raw (r"")

Sempre prefira escrever regex assim:

```python
r"\d+"
```

Ao invés de

```python
"\\d+"
```

O prefixo **r** impede que o Python interprete as barras invertidas.

São caracteres com significado especial.

```python
"[abc]"
```

```
a
b
c
```

Intervalo

Qualquer letra minúscula.

Maiúsculas.

Números.

Negação

```python
"[^abc]"
```

---

---

```python
Brasil$
```

---

# Sequências especiais

## \d

Dígito.

```python
\d
```

Equivale a

```
0-9
```

---

## \D

Não é dígito.

---

## \w

Caracteres de palavra.

Inclui

- letras
- números
- _

---

## \W

Tudo que NÃO é palavra.

---

## \s

Espaço em branco.

Inclui

- espaço
- tab
- quebra de linha

---

## \S

Não é espaço.

---

## \b

Limite de palavra.

```python
r"\bcat"
```

Encontra

```
cat
```

mas não

```
educate
```

---

## \B

Não está no limite da palavra.

---

## \A

Início absoluto da string.

---

## \Z

Fim absoluto da string.

---

# Exemplos práticos

## Encontrar números

```python
texto = "Tenho 20 anos e 3 gatos"

re.findall(r"\d+", texto)
```

Resultado

```
['20', '3']
```

---

## Encontrar palavras

```python
re.findall(r"\w+", texto)
```

---

## Encontrar espaços

```python
re.findall(r"\s", texto)
```

---

## Encontrar palavras iniciadas com A

```python
r"\bA\w+"
```

---

## Validar CEP

```
12345-678
```

Regex

```python
r"\d{5}-\d{3}"
```

---

## Validar telefone

```
(47)99999-9999
```

Regex

```python
r"\(\d{2}\)\d{5}-\d{4}"
```

---

## Encontrar extensão de arquivo

```
foto.jpg
```

Regex

```python
r"\.jpg$"
```

---

## Encontrar PDF

```python
r"\.pdf$"
```

---

## Encontrar qualquer imagem

```python
r"\.(jpg|png|jpeg|gif)$"
```

# Dica importante

Quase toda regex em Python deve ser escrita usando **Raw String**:

```python
r"\d+"
```

e não

```python
"\\d+"
```

Isso evita problemas com a barra invertida (`\`).
