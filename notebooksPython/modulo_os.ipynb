# O que é o módulo os?

O módulo **os** permite que o Python interaja diretamente com o Sistema Operacional.

Com ele podemos:

- navegar entre pastas;
- criar e excluir arquivos;
- criar e excluir diretórios;
- renomear arquivos;
- listar arquivos;
- descobrir caminhos;
- acessar variáveis do sistema;
- executar comandos do sistema.

É um dos módulos mais utilizados em automação.

# Percorrer uma árvore inteira

```python
for raiz, pastas, arquivos in os.walk("Fotos"):

    print(raiz)
    print(pastas)
    print(arquivos)
```

# Exemplo prático 1

Renomear todas as imagens

```python
import os

arquivos = os.listdir()

contador = 1

for arquivo in arquivos:

    if arquivo.endswith(".jpg"):

        novo = f"Imagem_{contador}.jpg"

        os.rename(arquivo, novo)

        contador += 1
```

---

# Exemplo prático 2

Encontrar todos os PDFs

```python
for arquivo in os.listdir():

    if arquivo.endswith(".pdf"):

        print(arquivo)
```

---

# Exemplo prático 3

Separar arquivos por extensão

```python
for arquivo in os.listdir():

    nome, extensao = os.path.splitext(arquivo)

    print(nome)

    print(extensao)
```

Resultado

```
foto .jpg

texto .pdf

planilha .xlsx
```

---

# Exemplo prático 4

Ler todas as subpastas

```python
for raiz, pastas, arquivos in os.walk("Downloads"):

    print(raiz)

    for arquivo in arquivos:

        print(arquivo)
```

# Quando usar o módulo os?

Use o `os` sempre que precisar interagir com o sistema operacional.

Casos comuns:

- Organizar arquivos automaticamente.
- Renomear milhares de arquivos.
- Criar backups.
- Encontrar PDFs ou imagens.
- Automatizar processos.
- Navegar entre diretórios.
- Desenvolver scripts para administração de sistemas.

---

# Observação importante

Embora o módulo `os` seja amplamente utilizado, para manipulação de caminhos modernos o Python recomenda o uso de **pathlib**, que oferece uma interface mais orientada a objetos e legível. Ainda assim, conhecer `os` é essencial, pois muitos projetos e bibliotecas utilizam esse módulo.
